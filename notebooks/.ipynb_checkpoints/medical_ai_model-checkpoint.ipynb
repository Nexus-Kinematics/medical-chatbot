```python
import pandas as pd
import numpy as np

# Work from the main dataset
split_df = df_clean[df_clean["diseases"].isin(stratifiable_classes)].copy()

# Lists for train/test indices
train_indices = []
test_indices = []

# Split EACH disease separately
for disease, group in split_df.groupby("diseases"):
    
    indices = group.index.to_numpy()
    
    # Reproducible shuffle
    rng = np.random.default_rng(42)
    rng.shuffle(indices)
    
    n = len(indices)
    
    # At least 1 record in test
    n_test = max(1, int(round(n * 0.20)))
    
    # At least 1 record must remain in training
    n_test = min(n_test, n - 1)
    
    test_indices.extend(indices[:n_test])
    train_indices.extend(indices[n_test:])

# Create final datasets
X_train = X_clean.loc[train_indices]
y_train = y_clean.loc[train_indices]

X_test = X_clean.loc[test_indices]
y_test = y_clean.loc[test_indices]

print("Split completed!")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
import pandas as pd

# Use the 727 diseases that have at least 2 records
split_df = df_clean[
    df_clean["diseases"].isin(stratifiable_classes)
].copy()

train_indices = []
test_indices = []

# Split each disease separately
for disease, group in split_df.groupby("diseases"):
    
    # Randomly select approximately 20% for testing
    n = len(group)
    
    n_test = max(1, round(n * 0.20))
    n_test = min(n_test, n - 1)
    
    test_part = group.sample(
        n=n_test,
        random_state=42
    )
    
    train_part = group.drop(test_part.index)
    
    test_indices.extend(test_part.index.tolist())
    train_indices.extend(train_part.index.tolist())

print("Splitting completed!")
print("Training indices:", len(train_indices))
print("Testing indices:", len(test_indices))
```

```python
import gc

# Remove temporary split variables if they exist
for var in [
    "split_df",
    "X_two",
    "y_two",
    "X_three_plus",
    "y_three_plus",
    "X_train_large",
    "X_test_large",
    "y_train_large",
    "y_test_large"
]:
    if var in globals():
        del globals()[var]

gc.collect()

print("Temporary variables cleared.")
```

```python
train_indices = []
test_indices = []

print("Index lists created.")
```

```python
for disease, group in y_clean.groupby(y_clean):
    
    # Ignore the 46 diseases with only one record
    if len(group) < 2:
        continue
    
    indices = group.index.tolist()
    
    # Randomly sample test indices using pandas
    n_test = max(1, round(len(indices) * 0.20))
    
    # Never put all records into test
    n_test = min(n_test, len(indices) - 1)
    
    test_sample = group.sample(
        n=n_test,
        random_state=42
    )
    
    test_idx = test_sample.index.tolist()
    
    train_idx = [i for i in indices if i not in set(test_idx)]
    
    test_indices.extend(test_idx)
    train_indices.extend(train_idx)

print("Splitting completed!")
print("Training indices:", len(train_indices))
print("Testing indices:", len(test_indices))
```
