```python
def predict_disease(test_symptoms):
    input_data = pd.DataFrame(
        0,
        index=[0],
        columns=symptom_names,
        dtype="uint8"
    )

    for symptom in test_symptoms:
        if symptom in input_data.columns:
            input_data.loc[0, symptom] = 1

    probabilities = model.predict_proba(input_data)[0]

    top_indices = probabilities.argsort()[-3:][::-1]

    results = []

    for index in top_indices:
        results.append({
            "disease": model.classes_[index],
            "probability": float(probabilities[index])
        })

    return results

print("Prediction function created!")
```

```python
def predict_top3(test_symptoms):
    valid_symptoms = [
        symptom for symptom in test_symptoms
        if symptom in symptom_names
    ]

    input_data = pd.DataFrame(
        0,
        index=[0],
        columns=symptom_names,
        dtype="uint8"
    )

    for symptom in valid_symptoms:
        input_data.loc[0, symptom] = 1

    probabilities = model.predict_proba(input_data)[0]

    top_indices = probabilities.argsort()[-3:][::-1]

    results = []

    for index in top_indices:
        results.append({
            "condition": model.classes_[index],
            "confidence": round(
                float(probabilities[index]) * 100,
                2
            )
        })

    return results
```
